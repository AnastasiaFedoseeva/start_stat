## **Задание 1**
Продукт считает, что среднее время на карточке вакансии равно 120 секунд. Собрали n = 100 пользователей, x = 124 секунды Разброс одного наблюдения в генеральной совокупности известен заранее: σ = 25 секунд. Сформулируйте H0. Посчитайте, на сколько стандартных ошибок оценка ушла от 120. Решите, есть ли основание отвергать H0.

Второй группы нет, есть эталон, заданный заранее. Генеральное $\sigma$ известно, поэтому применяем z-тест.

Метрика - среднее время на карточке, единица наблюдения - пользователь.

Гипотезы:

$H_0:\ \mu$ = 120

$H_1:\ \mu \ne$ 120

Отклонение оценки от гипотезы:
$\bar{x} - \mu_0$ = 124 - 120 = 4 с

Стандартная ошибка среднего:
$SE(\bar{x}) = \frac{\sigma}{\sqrt{n}} = \frac{25}{\sqrt{100}} = \frac{25}{10} = 2{,}5$ с

Z-статистика: на сколько стандартных ошибок оценка ушла от гипотезы:
$z = \frac{\bar{x} - \mu_0}{SE(\bar{x})} = \frac{4}{2{,}5} = 1{,}6$

Порог для отклонения $H_0$ из стандартного нормального закона при $\alpha = 0{,}05$ это $1{,}96$. Если |z| > 1,96, то отвергаем $H_0$, иначе оснований отвергать нет.

|z| = 1,6 < 1,96

Вывод: оценка ушла от 120 с на 1,6 стандартной ошибки, оснований отвергать $H_0$ нет.

## **Задание 2**
Первая группа: n1 = 40, x1 = 50, выборочное среднеквадратическое отклонение s1 = 10. Вторая группа: n2 = 40, x2 = 55, s2 = 10. Генеральные разбросы двух групп неизвестны и равными их считать не нужно. Посчитайте разницу средних и 95% доверительный интервал для неё. Напишите, попадает ли 0 в интервал и какой самый маленький эффект ещё согласуется с данными (который не будет опровергать H0).

Генеральное $\sigma$ неизвестно, есть только выборочное s, а равными разбросы считать не нужно, поэтому применяем Welch t-test.

Гипотезы:

$H_0:\ \mu_2 - \mu_1 = 0$

$H_1:\ \mu_2 - \mu_1 \ne 0$

Оценка эффекта - разница средних:

$\hat\Delta = \bar{x}_2 - \bar{x}_1$ = 55 - 50 = 5

Стандартная ошибка разницы (s - выборочное СКО, подставляем его):

$SE(\hat\Delta) = \sqrt{\frac{s_2^2}{n_2} + \frac{s_1^2}{n_1}} = \sqrt{\frac{10^2}{40} + \frac{10^2}{40}} = \sqrt{2{,}5 + 2{,}5} = \sqrt{5} \approx 2{,}236$

Доверительный интервал:
$\hat\Delta \pm t_{1-\alpha/2,\,\nu}\cdot SE(\hat\Delta)$

Для 95% берём $\alpha$ = 0,05. При большом n типично $t \approx 1{,}96$. Выборки у нас по 40 наблюдений, поэтому берём $t \approx 1{,}96$ как приближение:

$5 \pm 1{,}96 \cdot 2{,}236 = 5 \pm 4{,}38 \ \Rightarrow\ [0{,}62;\ 9{,}38]$

Вывод:
- Разница средних $\hat\Delta = +5$.
- 95% доверительный интервал: $\approx$ [0,62; 9,38].
- 0 в интервал не попадает.
- Самый маленький эффект, который ещё согласуется с данными, - нижняя граница интервала, примерно +0,6. Любое число внутри интервала, поставленное в $H_0$, не будет отвергнуто, а левее нижней границы уже будет.

## **Задание 3**

Первая группа: 90 откликов на 1000 показов карточки. Вторая группа: 120 откликов на 1000 показов. Каждый показ относится к своему пользователю, группы независимы. Сформулируйте H0 про две конверсии. Посчитайте разницу в процентных пунктах и статистику критерия для двух долей. Решите, есть ли основание отвергать H0.

Конверсия - это доля успехов: пользователь совершил целевое действие или нет. Исход бинарный, наблюдения независимы, поэтому применяем z-тест для разницы двух пропорций.

Гипотезы:

$H_0:\ p_1 = p_2$ (то есть $p_2 - p_1 = 0$)

$H_1:\ p_1 \ne p_2$

Оценка эффекта:

$\hat{p}_1 = \frac{90}{1000} = 0{,}09 = 9\%$

$\hat{p}_2 = \frac{120}{1000} = 0{,}12 = 12\%$

$\hat{p}_2 - \hat{p}_1 = 0{,}12 - 0{,}09 = 0{,}03 = +3$ п.п.

Если $H_0$ верна, у обеих групп одна и та же конверсия:

$\hat{p} = \frac{x_1 + x_2}{n_1 + n_2} = \frac{90 + 120}{1000 + 1000} = \frac{210}{2000} = 0{,}105$

Стандартная ошибка:
$SE = \sqrt{\hat{p}(1-\hat{p})\left(\frac{1}{n_2} + \frac{1}{n_1}\right)} = \sqrt{0{,}105 \cdot 0{,}895 \cdot \left(\frac{1}{1000} + \frac{1}{1000}\right)}$

Считаем по частям:

0,105 * 0,895 = 0,093975

$\ \frac{1}{1000} + \frac{1}{1000} = 0{,}002$

0,093975 * 0,002 = 0,00018795

Тогда
$SE = \sqrt{0{,}00018795} \approx 0{,}0137$

Статистика критерия:

$z = \frac{\hat{p}_2 - \hat{p}_1}{SE} = \frac{0{,}03}{0{,}0137} \approx 2{,}19$

Порог при $\alpha$ = 0,05 - 1,96: |z| = 2,19 > 1,96.

Вывод: есть основание отвергнуть $H_0$.

## **Задание 4**
Повторите эксперимент 2000 раз. В каждом повторе заново наберите две
группы по n = 200 наблюдений из одного и того же N (0, 1). На каждой паре групп посчитайте t Уэлча и его p-value. Нарисуйте распределение p-value. Напишите,какую форму оно имеет и почему.


Стартовое число генератора случайных чисел для заданий 4-7: 8.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind, mannwhitneyu

ALPHA = 0.05   # уровень значимости
```

```python
rng = np.random.default_rng(8)      # стартовое число 8
n_repeats, n = 2000, 200

t_stats = np.empty(n_repeats)
p_values = np.empty(n_repeats)
for i in range(n_repeats):
    group_1 = rng.normal(0, 1, n)
    group_2 = rng.normal(0, 1, n)
    welch = ttest_ind(group_1, group_2, equal_var=False)
    t_stats[i], p_values[i] = welch.statistic, welch.pvalue

plt.hist(p_values, bins=10, range=(0, 1), edgecolor="white")
plt.xlabel("p-value")
plt.ylabel("число повторов")
plt.title("Распределение p-value по 2000 повторам")
plt.tight_layout()
plt.show()
```

Вывод: распределение p-value равномерное на отрезке от 0 до 1, горки у нуля нет.

Причина в том, что обе группы каждый раз берутся из одного и того же N(0, 1), то есть $H_0$ верна по построению. При верной $H_0$ p-value равномерно распределён на отрезке от 0 до 1, поэтому при $\alpha$ = 0,05 ложные срабатывания, когда p-value <= 0,05, возникают примерно в 5% повторов.

## **Задание 5**

Повторите эксперимент 1000 раз. Первая группа каждый раз из N (0, 1), n = 60. Вторая из N (1, 1), тоже n = 60 после этого два её наблюдения замените на −50. На каждой паре групп посчитайте разницу средних, t Уэлча, U Манна–Уитни для первой группы и p-value обоих критериев. Нарисуйте распределения t и U. Нарисуйте распределения p-value обоих критериев и отметьте на них 0,05. Сравните, где они сосредоточены. Какое решение при двустороннем пороге 5% принимает каждый критерий? Совпали выводы или разошлись? Объясните, почему.

```python
rng = np.random.default_rng(8)      # стартовое число 8
n_repeats, n = 1000, 60

diffs, t_stats, u_stats, p_welch, p_mw, sd_2 = [], [], [], [], [], []
for _ in range(n_repeats):
    group_1 = rng.normal(0, 1, n)
    group_2 = rng.normal(1, 1, n)
    group_2[:2] = -50

    diffs.append(group_2.mean() - group_1.mean())
    welch = ttest_ind(group_2, group_1, equal_var=False)
    t_stats.append(welch.statistic)
    p_welch.append(welch.pvalue)
    mw = mannwhitneyu(group_1, group_2, alternative="two-sided")
    u_stats.append(mw.statistic)
    p_mw.append(mw.pvalue)
    sd_2.append(group_2.std(ddof=1))

diffs, t_stats, u_stats = np.array(diffs), np.array(t_stats), np.array(u_stats)
p_welch, p_mw, sd_2 = np.array(p_welch), np.array(p_mw), np.array(sd_2)

print(f"Разница средних: в среднем {diffs.mean():.3f}; доля повторов с разницей < 0: {(diffs < 0).mean():.3f}")
print(f"Выборочное СКО группы 2 в среднем: {sd_2.mean():.2f}")
print(f"t Уэлча: в среднем {t_stats.mean():.3f}; U (1-я группа): в среднем {u_stats.mean():.0f}, при H0 ожидали бы {n * n / 2:.0f}")
print(f"Медиана p-value: Welch = {np.median(p_welch):.3f}; Манн-Уитни = {np.median(p_mw):.2e}")

reject_w, reject_m = p_welch <= ALPHA, p_mw <= ALPHA
print(f"Доля повторов, где H0 отвергнута при alpha = {ALPHA}: Welch = {reject_w.mean():.3f}; Манн-Уитни = {reject_m.mean():.3f}")
print(f"Оба отвергли H0: {(reject_w & reject_m).sum()}")
print(f"Только Манн-Уитни отверг H0: {(~reject_w & reject_m).sum()}")
print(f"Только Welch отверг H0: {(reject_w & ~reject_m).sum()}")
print(f"Оба не отвергли H0: {(~reject_w & ~reject_m).sum()}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.6))
axes[0].hist(t_stats, bins=40, edgecolor="white")
axes[0].set_xlabel("t Уэлча"); axes[0].set_ylabel("число повторов"); axes[0].set_title("Распределение t")

axes[1].hist(u_stats, bins=40, edgecolor="white")
axes[1].set_xlabel("U (первая группа)"); axes[1].set_title("Распределение U")
plt.tight_layout(); plt.show()

fig, axes = plt.subplots(1, 2, figsize=(11, 3.6), sharey=True)
for ax, pv, name in [(axes[0], p_welch, "Welch t-test"), (axes[1], p_mw, "Манн–Уитни")]:
    ax.hist(pv, bins=20, range=(0, 1), edgecolor="white")
    ax.axvline(ALPHA, color="black", linestyle="--", linewidth=1.4, label="порог 0,05")
    ax.set_xlabel("p-value"); ax.set_title(f"p-value: {name}"); ax.legend()
axes[0].set_ylabel("число повторов")
plt.tight_layout()
plt.show()
```

Вывод:
- Разница средних во всех 1000 повторах отрицательная (в среднем $\approx$ -0,70), хотя у 58 из 60 пользователей группы 2 значения на +1 выше. Причина в том, что два значения -50 тянут среднюю группы 2 вниз: (58 * 1 + 2 * (-50))/60 = -0,7.
- Welch: t в среднем $\approx$ -0,58, то есть по модулю заметно меньше порога около 2. p-value далеко правее порога 0,05 (медиана $\approx$ 0,56). Выбросы раздувают выборочное СКО группы 2 примерно до 9 вместо 1, поэтому стандартная ошибка разницы $\sqrt{s_2^2/n_2 + s_1^2/n_1}$ получается 1,2, и сдвиг средних $\approx$ -0,7 не отличим от шума. Решение: $H_0$ не отвергается ни в одном из 1000 повторов.
- Манн–Уитни: U первой группы в среднем около 957 при ожидаемых по $H_0$ 1800 (то есть первая группа выше второй лишь примерно в 27% пар), p-value сжаты у нуля (медиана $\approx 9\cdot 10^{-6}$). $H_0$ отвергается в 999 из 1000 повторов.

Таким образом, выводы разошлись: в 999 из 1000 повторов Манн–Уитни отверг $H_0$, а Welch нет. В одном повторе $H_0$ не отвергли оба критерия.

Это связано с тем, что Welch работает со средним и разбросом, а они оба чувствительны к выбросам. Два значения -50 портят и оценку эффекта, и её шум. Манн–Уитни же работает с рангами, и значение -50 получает просто наименьший ранг. Выбросы стоят ему всего двух рангов, а сдвиг остальных 58 значений вверх виден в большинстве пар, поэтому различие распределений обнаруживается уверенно.

Критерии при этом отвечают на разные вопросы. Манн–Уитни не проверяет среднее. Его отклонение $H_0$ означает, что распределения различаются, а не то, что различаются средние.

## **Задание 6**
Один раз наберите первую группу из N (10, 2), n1 = 80, и вторую из N (10,8, 2,6), n2 = 75. Посчитайте 95% доверительный интервал Уэлча для разницы средних «вторая минус первая» и p-value критерия Уэлча. Решите, есть ли основание отвергать H0 при двустороннем пороге 5%. Напишите, попадает ли 0 в интервал и какой самый маленький эффект ещё согласуется с данными (не отвергает H0).

```python
rng = np.random.default_rng(8)          # стартовое число 8
n1, n2 = 80, 75
group_1 = rng.normal(10, 2, n1)
group_2 = rng.normal(10.8, 2.6, n2)

result = ttest_ind(group_2, group_1, equal_var=False, alternative="two-sided")
ci = result.confidence_interval(confidence_level=0.95)

print(f"95% доверительный интервал: [{ci.low:.3f}, {ci.high:.3f}]")
print(f"p-value = {result.pvalue:.4f}")
```

### Вывод к задаче 6

- 95% доверительный интервал Уэлча для разницы «вторая минус первая»: [-0,048; 1,430].
- p-value $\approx$ 0,0665.
- Основания отвергать $H_0$ нет: p = 0,0665 > 0,05 при двустороннем пороге 5%.
- 0 в интервал попадает, то есть вывод подтверждается.
- Самый маленький эффект, который ещё согласуется с данными - нижняя граница, $\approx$ -0,05. Значения эффекта от -0,05 до +1,43, в том числе 0, не отвергаются.

## **Задача 7**

Возьмите те же две группы, что в предыдущей задаче. 2000 раз заново
соберите выборки того же размера: каждый раз берите наблюдения из уже
полученных групп с возвращением (в numpy для этого есть rng.choice с
 параметром replace=True) и заново считайте разницу средних. Нарисуйте
распределение этих разниц. 95% интервал бутстрапа - границы, отсекающие по 2,5% значений с каждого края. Сравните эти границы с интервалом из предыдущей задачи. Отдельно посчитайте, в какой доле таких повторных выборок разница получилась не больше нуля. Можно ли эту долю назвать p-value? Объясните.

```python
rng = np.random.default_rng(8)          # стартовое число 8
n_boot = 2000

boot_diffs = np.empty(n_boot)
for i in range(n_boot):
    boot_1 = rng.choice(group_1, size=len(group_1), replace=True)
    boot_2 = rng.choice(group_2, size=len(group_2), replace=True)
    boot_diffs[i] = boot_2.mean() - boot_1.mean()

ci_boot_low, ci_boot_high = np.percentile(boot_diffs, [2.5, 97.5])
share_le_zero = (boot_diffs <= 0).mean()

print(f"Наблюдаемая разница в исходных группах: {group_2.mean() - group_1.mean():.3f}")
print(f"Центр бутстрап-распределения (среднее разниц): {boot_diffs.mean():.3f}")
print(f"95% бутстрап-интервал: [{ci_boot_low:.3f}, {ci_boot_high:.3f}]")
print(f"95% интервал Уэлча (задача 6): [{ci.low:.3f}, {ci.high:.3f}]")
print(f"Доля повторов с разницей <= 0: {share_le_zero:.4f} ({int(share_le_zero * n_boot)} из {n_boot})")

plt.figure(figsize=(7.2, 3.6))
plt.hist(boot_diffs, bins=40, edgecolor="white")
plt.axvline(ci_boot_low, color="black", linestyle="--", linewidth=1.4)
plt.axvline(ci_boot_high, color="black", linestyle="--", linewidth=1.4, label="95% бутстрап-интервал")
plt.axvline(0, color="gray", linestyle=":", linewidth=1.4, label="0")
plt.xlabel("бутстрап-оценка разницы средних (группа 2 - группа 1)")
plt.ylabel("частота")
plt.legend()
plt.tight_layout(); plt.show()
```

### Вывод к задаче 7

Сравнение интервалов:

Бутстрап-интервал [-0,062; 1,431] почти совпадает с интервалом Уэлча [-0,048; 1,430]: нижние границы отличаются на 0,014, верхние - на 0,001. Небольшие расхождения естественны: бутстрап - случайная процедура, 2000 повторов. 0 попадает в оба интервала.

Доля повторов с разницей <= 0 - около 0,037 (74 из 2000). Её нельзя назвать p-value по следующим причинам:

1. Не то распределение. P-value в бутстрапе - это доля повторов, где разница получилась не меньше наблюдаемой, если верна $H_0$, то есть распределение должно описывать мир, где различий нет, и быть центрировано на нуле. А мы берём выборки из исходных групп, где наблюдаемая разница равна 0,691, и расспределение центрировано около неё, а не около 0. Оно показывает разброс оценки, но не поведение разницы при верной $H_0$. Доля повторов <= 0 - это то, насколько далеко ноль от центра распределения, а не вероятность получить наблюдаемую или более экстремальную разницу при $H_0$.
2. Один хвост. Мы считаем только повторы с разницей <= 0 - одно направление. Двусторонняя проверка учитывает оба направления.
